```python
import json
from pathlib import Path
from sklearn.cluster import KMeans
import numpy as np

```

```python
# TODO: Load metadata
metadata_path = Path("../storage/object_store/eo_metadata.json")

with open(metadata_path, "r") as f:
    metadata = json.load(f)

print(f"Loaded {len(metadata)} products")
print(metadata[0])
```

```python
# TODO: implement clustering
def run_clustering(metadata, K=3):
    features = np.array([[item["brightness"], item["contrast"]] for item in metadata])

    kmeans = KMeans(n_clusters=K, random_state=42)
    labels = kmeans.fit_predict(features)

    for i, item in enumerate(metadata):
        item["cluster"] = int(labels[i])

    return metadata

metadata = run_clustering(metadata)
print(metadata[0])
```
